```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_excel("../data/raw/Online Retail.xlsx")
print (df.head())
```

```python
row, col = df.shape

print ("Total Rows: ", row)
print ("Total Columns: ", col)
print ("Data Type:\n", df.dtypes)
print ("Data of first 10 records:\n", df.head(10))
```

```python
# ==================================================
# 1. Convert DataFrame Columns to NumPy Arrays
# ==================================================

quantity_arr = df["Quantity"].to_numpy()
unitprice_arr = df["UnitPrice"].to_numpy()
country_arr = df["Country"].to_numpy()

# ==================================================
# 2. Array Information
# ==================================================

print("========== ARRAY INFORMATION ==========")

print(f"Quantity Array Shape : {quantity_arr.shape}")
print(f"UnitPrice Array Shape: {unitprice_arr.shape}")

print(f"Quantity Data Type   : {quantity_arr.dtype}")
print(f"UnitPrice Data Type  : {unitprice_arr.dtype}")

# ==================================================
# 3. Quantity Statistics
# ==================================================

print("\n========== QUANTITY STATISTICS ==========")

print(f"Minimum Quantity : {np.min(quantity_arr)}")
print(f"Maximum Quantity : {np.max(quantity_arr)}")
print(f"Mean Quantity    : {np.mean(quantity_arr):.2f}")
print(f"Median Quantity  : {np.median(quantity_arr):.2f}")

# ==================================================
# 4. Unit Price Statistics
# ==================================================

print("\n========== UNIT PRICE STATISTICS ==========")

print(f"Minimum Unit Price : {np.min(unitprice_arr):.2f}")
print(f"Maximum Unit Price : {np.max(unitprice_arr):.2f}")
print(f"Mean Unit Price    : {np.mean(unitprice_arr):.2f}")
print(f"Median Unit Price  : {np.median(unitprice_arr):.2f}")

# ==================================================
# 5. Country Information
# ==================================================

unique_countries = np.unique(country_arr)

print("\n========== COUNTRY INFORMATION ==========")

print(f"Total Unique Countries: {len(unique_countries)}")
print("Countries:", unique_countries)
```

```python

# Convert DataFrame columns to NumPy arrays
quantity_arr = df["Quantity"].to_numpy()
unitprice_arr = df["UnitPrice"].to_numpy()
customer_id_arr = df["CustomerID"].to_numpy()
invoice_arr = df["InvoiceNo"].to_numpy().astype(str)
stockcode_arr = df["StockCode"].to_numpy().astype(str)
country_arr = df["Country"].to_numpy()

# ==================================================
#  NEGATIVE QUANTITIES
# ==================================================

negative_quantity = np.sum(quantity_arr < 0)

print("Negative Quantities:", negative_quantity)


# ==================================================
#  ZERO QUANTITIES
# ==================================================

zero_quantity = np.sum(quantity_arr == 0)

print("Zero Quantities:", zero_quantity)


# ==================================================
#  NEGATIVE UNIT PRICES
# ==================================================

negative_unitprice = np.sum(unitprice_arr < 0)

print("Negative Unit Prices:", negative_unitprice)


# ==================================================
#  ZERO UNIT PRICES
# ==================================================

zero_unitprice = np.sum(unitprice_arr == 0)

print("Zero Unit Prices:", zero_unitprice)


# ==================================================
#  MISSING CUSTOMER IDs
# ==================================================

missing_customer_mask = np.isnan(customer_id_arr)

missing_customers = np.sum(missing_customer_mask)

print("Missing Customer IDs:", missing_customers)


# ==================================================
#  CANCELLED INVOICE ROWS
# ==================================================

cancelled_mask = np.char.startswith(invoice_arr, "C")

cancelled_rows = np.sum(cancelled_mask)

unique_cancelled_invoices = np.unique(
    invoice_arr[cancelled_mask]
)

print("Cancelled Invoice Rows:", cancelled_rows)

print("Unique Cancelled Invoices:", len(unique_cancelled_invoices))


# ==================================================
#  VALID QUANTITY RECORDS
# ==================================================

valid_quantity_mask = quantity_arr > 0

valid_quantity_count = np.sum(valid_quantity_mask)

print("Valid Quantity Records:", valid_quantity_count)


# ==================================================
#  UNIQUE PRODUCTS
# ==================================================

unique_products = np.unique(stockcode_arr)

print("Total Unique Products:", len(unique_products))


# ==================================================
#  COUNTRY-WISE TRANSACTION RECORDS
# ==================================================

countries, counts = np.unique(
    country_arr,
    return_counts=True
)

print("\n========== COUNTRY-WISE TRANSACTIONS ==========")

for country, count in zip(countries, counts):
    print(f"{country:<25} : {count:,}")


# ==================================================
# FINAL DATA QUALITY REPORT
# ==================================================

print("\n========== DATA QUALITY REPORT ==========")

print(f"Total Dataset Records   : {len(quantity_arr):,}")
print(f"Negative Quantities     : {negative_quantity:,}")
print(f"Zero Quantities         : {zero_quantity:,}")
print(f"Negative Unit Prices    : {negative_unitprice:,}")
print(f"Zero Unit Prices        : {zero_unitprice:,}")
print(f"Missing Customer IDs    : {missing_customers:,}")
print(f"Cancelled Invoice Rows  : {cancelled_rows:,}")
print(f"Valid Quantity Records  : {valid_quantity_count:,}")
print(f"Unique Products         : {len(unique_products):,}")
print(f"Total Countries         : {len(countries):,}")
```

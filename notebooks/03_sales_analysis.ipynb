```python
# ==================================================
# LOAD CLEANED SALES DATA
# ==================================================

import numpy as np

# Load updated NPZ file
sales_data = np.load("../data/processed/cleaned_sales.npz")

# Load required arrays
quantity_arr = sales_data["quantity"]
unitprice_arr = sales_data["unitprice"]
stockcode_arr = sales_data["stockcode"]
country_arr = sales_data["country"]

# Calculate revenue
revenue = quantity_arr * unitprice_arr
total_revenue = np.sum(revenue)

# Verify loaded data
print("========== DATA LOADING REPORT ==========")
print("Available Arrays:", sales_data.files)
print("Total Records:", len(quantity_arr))
print("Country Data Type:", country_arr.dtype)
print("Total Countries:", len(np.unique(country_arr)))

print("Dataset loaded successfully!")
```

```python
# ==========================================
#  SALES REVENUE ANALYSIS
# ==========================================

print ("Quantity Array", quantity_arr)
print ("UnitPrice Array", unitprice_arr)
print()
# Calculate revenue for each transaction line
revenue = quantity_arr * unitprice_arr

# Calculate total revenue
total_revenue = np.sum(revenue)

# Display results
print("========== SALES REVENUE REPORT ==========")

print(f"Total Transactions: {len(revenue):,}")
print(f"Total Revenue: £{total_revenue:,.2f}")
```

```python
# ==========================================
# Calculate average revenue per transaction line
# ==========================================

avg_revenue = np.mean(revenue)

print(f"Average Revenue Per Transaction Line: £{avg_revenue:,.2f}")
```

```python
# ==================================================
# TOP 10 PRODUCTS BY REVENUE
# ==================================================

# Load product codes
stockcode_arr = sales_data["stockcode"]

# Get unique products
unique_products = np.unique(stockcode_arr)

print(f"Total Unique Products: {len(unique_products):,}")

# Calculate revenue for each product
product_revenue = []

for product in unique_products:
    product_mask = stockcode_arr == product
    total = np.sum(revenue[product_mask])
    product_revenue.append(total)

# Convert to NumPy array
product_revenue = np.array(product_revenue)

# Get indices of top 10 highest-revenue products
top_10_indices = np.argsort(product_revenue)[-10:][::-1]

# Display results
print("\n========== TOP 10 PRODUCTS BY REVENUE ==========")

for rank, index in enumerate(top_10_indices, start=1):
    print(
        f"{rank:2}. Product: {unique_products[index]:<10} "
        f"| Revenue: £{product_revenue[index]:,.2f}"
    )
```

```python
# ==================================================
# TOP 10 PRODUCTS BY QUANTITY SOLD
# ==================================================

# Load product codes
stockcode_arr = sales_data["stockcode"]

# Get unique products
unique_products = np.unique(stockcode_arr)

print(f"Total Unique Products: {len(unique_products):,}")

# Calculate total quantity sold for each product
product_quantity = []

for product in unique_products:
    product_mask = stockcode_arr == product
    total = np.sum(quantity_arr[product_mask])
    product_quantity.append(total)

# Convert to NumPy array
product_quantity = np.array(product_quantity)

# Get top 10 highest-quantity products
top_10_indices = np.argsort(product_quantity)[-10:][::-1]

# Display results
print("\n========== TOP 10 PRODUCTS BY QUANTITY ==========")

for rank, index in enumerate(top_10_indices, start=1):
    print(
        f"{rank:2}. Product: {unique_products[index]:<10} "
        f"| Quantity Sold: {product_quantity[index]:,}"
    )
```

```python
# ==================================================
#   MONTHLY SALES ANALYSIS
# ==================================================

# 1. Load invoice dates from cleaned dataset
invoice_date_arr = sales_data["invoice_data"]

# 2. Convert dates into year-month format
months = invoice_date_arr.astype("datetime64[M]")

# 3. Identify unique months
unique_months = np.unique(months)

print(f"Total Unique Months: {len(unique_months)}")

# ==================================================
#  CALCULATE MONTHLY REVENUE
# ==================================================

monthly_revenue = []

for month in unique_months:

    # Select transactions belonging to this month
    month_mask = months == month

    # Calculate total revenue for the month
    total = np.sum(revenue[month_mask])

    # Store monthly revenue
    monthly_revenue.append(total)

# Convert list into NumPy array
monthly_revenue = np.array(monthly_revenue)

# ==================================================
#  DISPLAY MONTHLY SALES REPORT
# ==================================================

print("\n========== MONTHLY SALES REPORT ==========")

for month, total in zip(unique_months, monthly_revenue):
    print(f"{month} | Revenue: £{total:,.2f}")

# ==================================================
#  MONTHLY SALES SUMMARY
# ==================================================

highest_month_index = np.argmax(monthly_revenue)
lowest_month_index = np.argmin(monthly_revenue)

print("\n========== MONTHLY SALES SUMMARY ==========")

print(f"Total Months: {len(unique_months)}")

print(
    f"Highest Revenue Month: {unique_months[highest_month_index]} "
    f"| £{monthly_revenue[highest_month_index]:,.2f}"
)

print(
    f"Lowest Revenue Month: {unique_months[lowest_month_index]} "
    f"| £{monthly_revenue[lowest_month_index]:,.2f}"
)

print(f"Total Revenue: £{np.sum(monthly_revenue):,.2f}")
```

```python
# Verify monthly revenue calculations
print(
    "Monthly Revenue Matches Total Revenue:",
    np.isclose(np.sum(monthly_revenue), total_revenue)
)
```

```python
# ==================================================
#  COUNTRY-WISE REVENUE ANALYSIS
# ==================================================

# 1. Load country data from cleaned dataset
country_arr = sales_data["country"]

# 2. Identify unique countries
unique_countries = np.unique(country_arr)

print(f"Total Unique Countries: {len(unique_countries)}")

# ==================================================
# CALCULATE REVENUE FOR EACH COUNTRY
# ==================================================

country_revenue = []

for country in unique_countries:

    # Select transactions belonging to this country
    country_mask = country_arr == country

    # Calculate total revenue
    total = np.sum(revenue[country_mask])

    # Store country revenue
    country_revenue.append(total)

# Convert list into NumPy array
country_revenue = np.array(country_revenue)

# ==================================================
# SORT COUNTRIES BY REVENUE
# ==================================================

sorted_indices = np.argsort(country_revenue)[::-1]

# ==================================================
# DISPLAY COUNTRY-WISE REVENUE
# ==================================================

print("\n========== COUNTRY-WISE REVENUE REPORT ==========")

for rank, index in enumerate(sorted_indices, start=1):

    print(
        f"{rank:2}. Country: {unique_countries[index]:<20} "
        f"| Revenue: £{country_revenue[index]:,.2f}"
    )

# ==================================================
# TOP 10 COUNTRIES BY REVENUE
# ==================================================

top_10_indices = sorted_indices[:10]

print("\n========== TOP 10 COUNTRIES BY REVENUE ==========")

for rank, index in enumerate(top_10_indices, start=1):

    print(
        f"{rank:2}. {unique_countries[index]:<20} "
        f"| £{country_revenue[index]:,.2f}"
    )

# ==================================================
# COUNTRY REVENUE SUMMARY
# ==================================================

total_country_revenue = np.sum(country_revenue)

print("\n========== COUNTRY REVENUE SUMMARY ==========")

print(f"Total Countries: {len(unique_countries)}")
print(f"Total Revenue: £{total_country_revenue:,.2f}")

# Verify country-wise revenue
print(
    "Revenue Verification:",
    np.isclose(total_country_revenue, total_revenue)
)
```

```python
import numpy as np

# ==================================================
#  CUSTOMER ANALYSIS USING NUMPY
# ==================================================

# Load cleaned dataset
sales_data = np.load("../data/processed/cleaned_sales.npz")

# Load required arrays
customer_id_arr = sales_data["customer_id"]
quantity_arr = sales_data["quantity"]
unitprice_arr = sales_data["unitprice"]
invoice_arr = sales_data["invoice"]

# Calculate revenue for each transaction line
revenue = quantity_arr * unitprice_arr

print("========== DATA LOADING REPORT ==========")

print(f"Total Records: {len(quantity_arr):,}")
print(f"CustomerID Data Type: {customer_id_arr.dtype}")
print(f"Invoice Data Type: {invoice_arr.dtype}")
print(f"Total Revenue: £{np.sum(revenue):,.2f}")
```

```python
# ==================================================
#  FILTER MISSING CUSTOMER IDs
# ==================================================

# Identify customers with missing IDs
missing_customer_mask = np.isnan(customer_id_arr)

# Select valid customer IDs
valid_customer_mask = ~missing_customer_mask

# Apply the same mask to related arrays
valid_customer_ids = customer_id_arr[valid_customer_mask]
valid_revenue = revenue[valid_customer_mask]
valid_invoices = invoice_arr[valid_customer_mask]

print("\n========== CUSTOMER DATA REPORT ==========")

print(f"Total Sales Records: {len(customer_id_arr):,}")
print(f"Missing Customer IDs: {np.sum(missing_customer_mask):,}")
print(f"Valid Customer Records: {len(valid_customer_ids):,}")
```

```python
# ==================================================
#  UNIQUE CUSTOMERS
# ==================================================

# Identify unique customers
unique_customers = np.unique(valid_customer_ids)

# Count customers
total_customers = len(unique_customers)

print("\n========== UNIQUE CUSTOMERS REPORT ==========")

print(f"Total Unique Customers: {total_customers:,}")
print("First 10 Customer IDs:", unique_customers[:10])
```

```python
# ==================================================
#  CUSTOMER-WISE TOTAL SPENDING
# ==================================================

customer_spending = []

for customer in unique_customers:

    # Identify transactions belonging to this customer
    customer_mask = valid_customer_ids == customer

    # Calculate total spending
    total = np.sum(valid_revenue[customer_mask])

    # Store spending
    customer_spending.append(total)

# Convert list into NumPy array
customer_spending = np.array(customer_spending)

print("\n========== CUSTOMER SPENDING REPORT ==========")

# Display first 10 customers
for customer, spending in zip(
    unique_customers[:10],
    customer_spending[:10]
):
    print(
        f"Customer: {int(customer)} "
        f"| Total Spending: £{spending:,.2f}"
    )
```

```python
# ==================================================
#  TOP 10 CUSTOMERS BY REVENUE
# ==================================================

# Sort customer spending in descending order
top_10_indices = np.argsort(customer_spending)[-10:][::-1]

print("\n========== TOP 10 CUSTOMERS BY REVENUE ==========")

for rank, index in enumerate(top_10_indices, start=1):

    print(
        f"{rank:2}. Customer ID: {int(unique_customers[index])} "
        f"| Total Spending: £{customer_spending[index]:,.2f}"
    )
```

```python
# ==================================================
#  TOP 10 CUSTOMERS BY REVENUE
# ==================================================

# Sort customer spending in descending order
top_10_indices = np.argsort(customer_spending)[-10:][::-1]

print("\n========== TOP 10 CUSTOMERS BY REVENUE ==========")

for rank, index in enumerate(top_10_indices, start=1):

    print(
        f"{rank:2}. Customer ID: {int(unique_customers[index])} "
        f"| Total Spending: £{customer_spending[index]:,.2f}"
    )
```

```python
# ==================================================
#  UNIQUE ORDERS PER CUSTOMER
# ==================================================

# Store each customer's total orders
customer_orders = []

for customer in unique_customers:

    # Find transactions belonging to this customer
    customer_mask = valid_customer_ids == customer

    # Select customer invoices
    invoices = valid_invoices[customer_mask]

    # Count unique orders
    total_orders = len(np.unique(invoices))

    # Store order count
    customer_orders.append(total_orders)

# Convert list into NumPy array
customer_orders = np.array(customer_orders)

print("========== CUSTOMER ORDERS REPORT ==========")

print(f"Total Customers: {len(customer_orders):,}")
print("First 10 Customer Order Counts:", customer_orders[:10])
```

```python
# ==================================================
#  AVERAGE ORDER VALUE (AOV)
# ==================================================

# Calculate average order value per customer
customer_aov = np.divide(
    customer_spending,
    customer_orders,
    out=np.zeros_like(customer_spending, dtype=float),
    where=customer_orders != 0
)

print("\n========== CUSTOMER AOV REPORT ==========")

for customer, aov in zip(
    unique_customers[:10],
    customer_aov[:10]
):
    print(
        f"Customer: {int(customer)} "
        f"| Average Order Value: £{aov:,.2f}"
    )

# Calculate overall average order value
total_unique_orders = len(np.unique(valid_invoices))

overall_aov = np.sum(valid_revenue) / total_unique_orders

print("\n========== OVERALL AOV ==========")

print(f"Total Unique Orders: {total_unique_orders:,}")
print(f"Overall AOV: £{overall_aov:,.2f}")
```

```python
# ==================================================
#  CUSTOMER SEGMENTATION
# ==================================================

# Calculate spending percentiles
low_threshold = np.percentile(customer_spending, 33)
high_threshold = np.percentile(customer_spending, 66)

# Assign spending segments
customer_segments = np.where(
    customer_spending <= low_threshold,
    "Low",
    np.where(
        customer_spending <= high_threshold,
        "Medium",
        "High"
    )
)

# Count customers in each segment
segments, segment_counts = np.unique(
    customer_segments,
    return_counts=True
)

print("\n========== CUSTOMER SEGMENTATION REPORT ==========")

print(f"Low Spending Threshold: £{low_threshold:,.2f}")
print(f"High Spending Threshold: £{high_threshold:,.2f}")

for segment, count in zip(segments, segment_counts):
    print(f"{segment} Spending Customers: {count:,}")
```

```python
# ==================================================
# FINAL CUSTOMER ANALYSIS REPORT
# ==================================================

print("\n" + "=" * 55)
print("        CUSTOMER INTELLIGENCE REPORT")
print("=" * 55)

print(f"Total Sales Records       : {len(customer_id_arr):,}")
print(f"Valid Customer Records    : {len(valid_customer_ids):,}")
print(f"Missing Customer Records  : {np.sum(missing_customer_mask):,}")

print("-" * 55)

print(f"Total Unique Customers    : {total_customers:,}")
print(f"Total Unique Orders       : {total_unique_orders:,}")

print("-" * 55)

print(f"Identified Customer Revenue: £{np.sum(valid_revenue):,.2f}")
print(f"Overall Average Order Value: £{overall_aov:,.2f}")

print("-" * 55)

print("CUSTOMER SEGMENTS")

for segment, count in zip(segments, segment_counts):
    print(f"{segment:<10}: {count:,}")

print("=" * 55)
```

```python
# ==================================================
# VERIFY CUSTOMER ANALYSIS
# ==================================================

# Verify customer spending total
spending_verified = np.isclose(
    np.sum(customer_spending),
    np.sum(valid_revenue)
)

# Verify customer count
customer_count_verified = (
    np.sum(segment_counts) == total_customers
)

# Verify order counts
orders_verified = (
    np.sum(customer_orders) >= total_unique_orders
)

print("\n========== VERIFICATION REPORT ==========")

print("Customer Spending Verified:", spending_verified)
print("Customer Segments Verified:", customer_count_verified)
print("Customer Orders Consistent:", orders_verified)
```
